### Dropping Null values

```python
#dropping rows that have null values
filterd_df=df.dropna(axis=0)
filterd_df
```

### Searching for duplicates

```python
print("Is there any duplicates? :",filterd_df.duplicated().any())
```

### Some infomative Statistics

```python
filterd_df.describe(exclude=[object])
```

### Highest Rating Top 3 Movies

```python
filterd_df.columns
```

```python
Highest = filterd_df.nlargest(n=3,  columns=["vote_count"])
sns.barplot(x = 'original_title', y = 'vote_count',data=Highest)
plt.title("Rating each Year")
plt.show()
```

### Lowest Rating Top 3 Movies

```python
Lowest = filterd_df.nsmallest(n=3,  columns=["vote_count"])
sns.barplot(x = 'original_title', y = 'vote_count',data=Lowest)
plt.title("Rating each Year")
plt.show()
```

### #of Movies released every year

```python
filterd_df.groupby('release_year').count()['id'].plot(xticks=np.arange(1960,2016,5))
sns.set(rc={'figure.figsize':(10,5)})
plt.title("Movies released each Year")
plt.show()
```

### Highest and Lowest runtime

```python
Highest = filterd_df.nlargest(n=3,  columns=["runtime"])
sns.barplot(x = 'original_title', y = 'runtime',data=Highest)
plt.title("Highest Runtime of movies")
plt.show()
```

```python
Lowest = filterd_df.nsmallest(n=3,  columns=["runtime"])
sns.barplot(x = 'original_title', y = 'runtime',data=Lowest)
plt.title("Lowest Runtime of movies")
plt.show()
```
